## 1. Conditional Statements

Conditional statements allow your program to make decisions and execute specific blocks of code depending on whether a condition is `True` or `False`.

### Syntax Structure:
```python
if condition1:
    # block 1
elif condition2:
    # block 2
else:
    # block 3
```

### Key Operators:
- **Comparison Operators**: `==` (equal), `!=` (not equal), `<` (less than), `>` (greater than), `<=` (less than or equal), `>=` (greater than or equal).
- **Logical Operators**: `and` (both true), `or` (at least one true), `not` (inverts truth value).

````python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80 and score < 90:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print("Score:", score, "| Grade:", grade)

# Short-hand conditional (Ternary Operator)
status = "Pass" if score >= 70 else "Fail"
print("Status:", status)
````

## 2. Loops & Control Statements

Loops are used to execute a block of code repeatedly.

### Types of Loops:
1. **`for` loops**: Best for iterating over a sequence (list, tuple, string, dictionary) or a predefined range.
2. **`while` loops**: Repeats code as long as a condition remains `True`. Careful to avoid infinite loops!

### Loop Control Statements:
- **`break`**: Terminates the loop prematurely and transfers execution to the statement following the loop.
- **`continue`**: Skips the rest of the current iteration and jumps to the next iteration of the loop.
- **`pass`**: A null statement that acts as a placeholder when syntactic structure is required but no code needs execution.

````python
print("--- For Loop (with range) ---")

# range(start, stop, step) - stop is exclusive
for i in range(1, 10, 3):
    print(i)
````

````python
print("\n--- While Loop & Break Statement ---")

count = 1
while count <= 100:
    if count == 10:
        print("Breaking loop at count:", count)
        break
    print("Count:", count)
    count += 1 # count = count + 1
````

````python
print("\n--- Continue Statement ---")
for num in range(1, 6):
    if num == 5:
        # Skip printing 3
        continue
    print("Number:", num)
````

## 3. Functions & Arguments

A **function** is a reusable block of code designed to perform a specific action. You define functions using the `def` keyword.

### Key Concepts:
- **Parameters**: Variables named in the function definition.
- **Arguments**: Values passed to the function when calling it.
- **Default Arguments**: Parameter values that are used if no argument is supplied during the function call.
- **Docstrings**: Tripled-quoted strings written immediately after the function definition to document its purpose.
- **`return` statement**: Exits a function and optionally returns a value to the caller. If omitted, the function returns `None`.
- **Variable-Length Arguments**:
  - `*args`: Captures arbitrary positional arguments as a **tuple**.
  - `**kwargs`: Captures arbitrary keyword arguments as a **dictionary**.

````python
def calculate_area(width, height=10):
    """
    Calculates the area of a rectangle.
    If height is not provided, defaults to 10.
    """
    area = width * height
    return area

# Call using positional arguments
print("Area (5, 6):", calculate_area(5, 6))
````

````python
# Call relying on default arguments
print("Area (5, default):", calculate_area(5))
````

````python
# Call using keyword arguments
print("Area (height=4, width=8):", calculate_area(height=4, width=8))
````

````python
# Accessing function docstring
print("Docstring:", calculate_area.__doc__)
````

## 4. Python Data Structures

Data structures are containers used to organize and store data collections. Python has four built-in collection structures.

### A. Lists
- **Properties**: Ordered, mutable (changeable), and allows duplicate elements.
- **Common methods**: `.append()`, `.extend()`, `.insert()`, `.remove()`, `.pop()`, `.sort()`, `.reverse()`.

````python
fruits = ["apple", "banana", "cherry"]
print("Initial List:", fruits)
````

````python
# Modifying an element
fruits[1] = "blueberry"
print("After modification:", fruits)
````

````python
# Appending elements
fruits.append("orange")
print("After append:", fruits)
````

````python
# Removing elements
popped_fruit = fruits.pop(0)  # Removes by index
print("Popped item:", popped_fruit, "| Current List:", fruits)
````

````python
fruits.remove("cherry")      # Removes by value
print("After remove:", fruits)
````

### B. Tuples
- **Properties**: Ordered, **immutable** (cannot be altered after creation), and allows duplicate elements.
- **Syntax**: Defined using parentheses `()`.
- **Use case**: Read-only datasets, returning multiple values from a function, storing fixed coordinates.

````python
point = (3, 5)
print("Point coordinates:", point)
print("X value:", point[0])
````

````python
# Attempting to modify a tuple will raise a TypeError
try:
    point[0] = 10
except TypeError as e:
    print("Error caught:", e)
````

````python
# Tuple packing and unpacking
student = ("Fahd", 25, "AI")
name, age, subject = student
print(f"Unpacked Student -> Name: {name}, Age: {age}, Subject: {subject}")
````

### C. Sets
- **Properties**: Unordered, unindexed, does **not** allow duplicate elements.
- **Syntax**: Defined using curly braces `{}` containing individual elements.
- **Operations**: Supports mathematical set operations like union, intersection, difference, and symmetric difference.

````python
set_a = {1, 2, 3, 4, 5, 5, 5}  # Note that duplicate 5 will be ignored
set_b = {4, 5, 6, 7, 8}

print("Set A:", set_a)
print("Set B:", set_b)

````

````python
# Mathematical Set Operations
print("Union (A | B):", set_a | set_b)                   # All unique elements
````

````python
print("Intersection (A & B):", set_a & set_b)            # Common elements
````

````python
print("Difference (A - B):", set_a - set_b)              # In A but not in B
````

````python
print("Symmetric Difference (A ^ B):", set_a ^ set_b)    # In A or B but not both
````

### D. Dictionaries
- **Properties**: Key-value pairs. Ordered (since Python 3.7), keys must be unique and immutable, values can be duplicated and of any type.
- **Syntax**: `{key1: value1, key2: value2}`.

````python
mentor_profile = {
    "name": "Fahd",
    "role": "Mentor",
    "topics": ["Python", "Data Science", "Machine Learning"]
}

print("Mentor Profile:", mentor_profile)
````

````python
# Accessing values using keys
print("Name:", mentor_profile["name"])
````

````python
del mentor_profile["name"]
print(mentor_profile)
````

````python
# Accessing using .get() (safely handles missing keys by returning None or a default value)
print("Organization:", mentor_profile.get("organization", "Not Specified"))
````

````python
# Adding/Updating values
mentor_profile["role"] = "Lead Mentor"
mentor_profile["experience_years"] = 2
print("Updated Profile:", mentor_profile)
````

````python
mentor_profile.items()
````

````python
# Iterating through key-value pairs
print("\n--- Profile Keys & Values ---")

for key, value in mentor_profile.items():
    print(f"{key.capitalize()}: {value}")
````

## 5. When and Why to Use Each Structure

Choosing the correct data structure can significantly affect code performance and readability.

| Structure | Mutability | Ordering | Duplicates Allowed | Primary Use Case |
| :--- | :--- | :--- | :--- | :--- |
| **List** | Mutable | Ordered | Yes | General-purpose collection of items where ordering matters and values change. |
| **Tuple** | Immutable | Ordered | Yes | Fixed records, write-protected database outputs, coordinates, function outputs. |
| **Set** | Mutable | Unordered | No | Membership testing, deduplication, performing mathematical set operations. |
| **Dictionary** | Mutable | Ordered (since 3.7) | Keys: No, Values: Yes | Mapping associations, JSON data configurations, fast key-based lookups. |

## 6. List Comprehensions

List comprehensions offer a shorter syntax to create a new list from an existing sequence or iterable.

### Syntax:
`new_list = [expression for item in iterable if condition]`

````python
# Task: Generate a list containing the squares of numbers from 1 to 5

# Classic approach using a for loop:
squares_loop = []
for x in range(1, 6):
    squares_loop.append(x ** 2)
print("Squares (Classic Loop):", squares_loop)
````

````python
# List comprehension approach:
squares_comp = [x ** 2 for x in range(1, 6)]
print("Squares (Comprehension):", squares_comp)
````

````python
# List comprehension with conditional filter (only squares of odd numbers)
odd_squares_comp = [x ** 2 for x in range(1, 6) if x % 2 == 0]
print("Odd Squares (Comprehension):", odd_squares_comp)
````
