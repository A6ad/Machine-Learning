# 6. Calculate month-wise average windspeed

# 7. Calculate month-wise highest meantemp

```python
climate['date'] = pd.to_datetime(climate['date'])
monthly_max_temp = climate.groupby(climate['date'].dt.to_period('M'))['meantemp'].max().reset_index()

print(monthly_max_temp)

```

# 9. A new column contains previous 10 days average meantemp

```python

climate['date'] = pd.to_datetime(climate['date'])
climate = climate.sort_values('date').reset_index(drop=True)
climate['10_day_avg_temp'] = climate['meantemp'].rolling(window=10).mean()

print(climate[['date', 'meantemp', '10_day_avg_temp']].head(15))
```
