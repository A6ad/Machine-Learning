# stacking and spliting

```python
d = np.arange(27).reshape(3,9)
e = np.arange(28,55).reshape(3,9)
```

# use hstack() and vstack()

```python
horizontal_array = np.hstack((d, e))
vertical_array = np.vstack((d, e))

print("Horizontal stack:")
print(horizontal_array)

print("Vertical stack:")
print(vertical_array)
```

# use hsplit() and vsplit()

```python
horizontal_split = np.hsplit(d, 3)
vertical_split = np.vsplit(d, 3)

print("Horizontal split:")
for part in horizontal_split:
    print(part)

print("Vertical split:")
for part in vertical_split:
    print(part)
```
