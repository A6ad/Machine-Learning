## 2. Python Syntax and Basic Operations

Python is highly regarded for its readability. It has key syntactical features you must remember:
- **No Semicolons**: Unlike C++ or Java, Python statements do not require a semicolon (`;`) at the end of lines.
- **Comments**: Use the `#` symbol for single-line comments. Python ignores everything written after `#` on that line.
- **Indentation**: Python uses whitespace indentation (4 spaces is the standard) to define the scope of control structures (loops, functions, classes). If indentation is incorrect, Python will raise an `IndentationError`.
- **Case Sensitivity**: Variable and function names are case-sensitive. `data` and `Data` are completely distinct identifiers.

```python
# This is a single-line comment. Python will ignore this.

print("Hello, ML BootCamp!")  # This statement prints a string to the console

# Demonstration of indentation
if True:
    print("This print statement is inside the 'if' block.")
    print("It runs because it is correctly indented with 4 spaces.")
```

## 3. Variables and Data Types

A **variable** is a container that stores data values in memory. 
- **Dynamic Typing**: In Python, you do not need to explicitly declare the data type of a variable. Python automatically determines the type based on the value assigned to it.
- **Re-assignment**: You can change the data type of a variable dynamically by assigning it a value of a different type.

### Variable Naming Rules:
1. Variable names must start with a letter (a-z, A-Z) or an underscore (`_`).
2. They cannot start with a number.
3. They can only contain letters, numbers, and underscores.
4. They cannot be any of Python's reserved keywords (e.g., `if`, `else`, `while`, `def`, `class`, `import`, `True`, `False`, `None`).

### Fundamental Data Types:
- **Integer (`int`)**: Whole numbers without decimals (e.g., `10`, `-3`).
- **Float (`float`)**: Real numbers containing decimal points (e.g., `3.14`, `-0.005`).
- **Complex (`complex`)**: Numbers with real and imaginary parts, where `j` represents $\sqrt{-1}$ (e.g., `2 + 5j`).
- **String (`str`)**: Character sequence enclosed in single (`'`), double (`"`), or triple (`'''` or `"""`) quotes.
- **Boolean (`bool`)**: Binary values representing logical states: `True` or `False`.

```python
# Assigning values to variables
x = 42                 # Integer
y = 3.14159            # Float
z = 1 + 2j             # Complex
name = "Fahd"          # String
is_active = True       # Boolean
```

```python
# Displaying variables and checking their types using type()
print("x =", x, "| Type:", type(x))
print("y =", y, "| Type:", type(y))
print("z =", z, "| Type:", type(z))
print("name =", name, "| Type:", type(name))
print("is_active =", is_active, "| Type:", type(is_active))
```

```python
# Dynamic typing check
dynamic_var = 100
print("Value:", dynamic_var, "| Type:", type(dynamic_var))
dynamic_var = "Now I am a AI"
print("Value:", dynamic_var, "| Type:", type(dynamic_var))
```

## 4. Input and Output Operations

- **Output (`print()`)**: Used to display outputs. You can separate values using commas, change the separator using the `sep` parameter, and change the ending character using the `end` parameter.
- **Input (`input()`)**: Used to capture user inputs. **Crucial detail**: `input()` always captures inputs as a **string** (`str`). If you need numeric values, you must convert the result explicitly (typecasting).

```python
# Formatting print output using sep and end
print("Apple", "Banana", "Orange", sep=", ")
print("First Line", end=" --- ")
print("Second Line on the same console row")

```

```python
# Capturing user input
user_input = input("Enter your favorite color: ")
print("Your favorite color is:", user_input)
```

```python
# Capturing numbers and typecasting
num1_str = input("Enter a number: ")
num2_str = input("Enter another number: ")

# Attempting mathematical operations on raw string input will concatenate them
concat_result = num1_str + num2_str
print("Result without casting (concatenation):", concat_result)

```

```python
# Correct approach: cast strings to integers or floats
num1 = float(num1_str)
num2 = float(num2_str)
math_result = num1 + num2
print("Result with float casting (addition):", math_result)
```

```python
num3 = int(input("Enter a number: "))
print("You entered:", num3, "| Type:", type(num3))
```

## 5. Basic Arithmetic Operations

Python supports all basic arithmetic operations. The operators are listed below:

| Operator | Operation | Description | Example |
| :--- | :--- | :--- | :--- |
| `+` | Addition | Adds two values | `a + b` |
| `-` | Subtraction | Subtracts second value from first | `a - b` |
| `*` | Multiplication | Multiplies two values | `a * b` |
| `/` | Division | Divides first value by second (always returns `float`) | `a / b` |
| `//` | Floor Division | Divides and rounds down to nearest integer | `a // b` |
| `%` | Modulo | Returns remainder of division | `a % b` |
| `**` | Exponentiation | Raises base to power | `a ** b` |

```python
a = 17
b = 5

print("a =", a, ", b =", b)
print("Addition (a + b) =", a + b)
print("Subtraction (a - b) =", a - b)
print("Multiplication (a * b) =", a * b)
print("Division (a / b) =", a / b)          # Note that this is a float result
```

```python
print("Floor Division (a // b) =", a // b)  # Rounds down to 3
print("Modulo (a % b) =", a % b)          # Remainder of 17 / 5 is 2
print("Exponentiation (a ** b) =", a ** b)  # 17 raised to the power of 5
```

## 6. String Operations

Strings are sequences of characters. Python offers powerful operations and methods to manipulate strings.

### Concatenation and Repetition
- **Concatenation (`+`)**: Joins strings together.
- **Repetition (`*`)**: Repeats a string a given number of times.

```python
word1 = "Data"
word2 = "Science"

# Concatenate strings
phrase = word1 + " " + word2
print("Concatenated phrase:", phrase)

# Repeat a string
divider = "-" * 30
print(divider)
```

### String Indexing and Slicing
- **Zero-based Indexing**: Python indexes strings from `0` (left-to-right). 
- **Negative Indexing**: Right-to-left indexing starts at `-1`.
- **Slicing**: Extract a portion of a string using `string[start:end:step]`. Note that the `end` index is **exclusive** (not included).

```python
text = "MachineLearning"

# Indexing examples
print("Character at index 0:", text[0])
print("Character at index 7:", text[7])
print("Last character (index -1):", text[-1])
print("Second to last character (index -2):", text[-2])

# Slicing examples
print("Slicing [0:7] (indices 0 to 6):", text[0:7])       # "Machine"
print("Slicing [7:15] (indices 7 to 14):", text[7:15])     # "Learning"
print("Slicing from start to index 7 [:7]:", text[:7])
print("Slicing from index 7 to end [7:]:", text[7:])
print("Slicing with step [::2] (every 2nd char):", text[::2])
print("Reversing string using slice [::-1]:", text[::-1])
```

### Essential String Methods
Strings are objects and come with many built-in methods:

```python
message = "   Welcome to Python programming!   "

# Remove leading/trailing whitespaces
cleaned = message.strip()
print("Stripped message:", f"'{cleaned}'")
```

```python
# Lowercase and Uppercase conversion
print("Lower:", cleaned.lower())
print("Upper:", cleaned.upper())
```

```python
# Checking properties
print("Does it start with 'Welcome'?:", cleaned.startswith("Welcome"))
print("Is the word alphabetic?:", "Python".isalpha())
print("Is '123' a digit string?:", "123".isdigit())
```

```python
# Replacement
replaced_message = cleaned.replace("Python", "Machine Learning")
print("Replaced:", replaced_message)
```

```python
# Split a string into a list by a delimiter
words_list = cleaned.split(" ")
print("List of words:", words_list)
```

```python
# Check substring occurrence count
print("Occurrences of 'o' in cleaned string:", cleaned.count("o"))
```
